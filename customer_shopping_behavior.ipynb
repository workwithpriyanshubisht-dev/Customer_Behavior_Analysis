```python
import pandas as pd
df = pd.read_csv('customer_shopping_behavior.csv')
df.head()
df.info()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns =  df.columns.str.replace(' ',"_")
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'}) 
```

```python
df.columns
```

```python
#create a coulmn age_group
labels = ['young adult','adult','middle-aged','senior']
df['age_group'] = pd.qcut(df['age'], q=4,labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create columns purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly':  30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
```

```python
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus

username = "root"
password = "priyanshu"
host = "127.0.0.1"
port = 3306
database = "customer_shopping_behavior"
old_table_name = "customer_shopping_behavior"
table_name = "customer_data"

authenticated_url = (
    f"mysql+pymysql://{quote_plus(username)}:{quote_plus(password)}"
    f"@{host}:{port}/{database}"
)
engine = create_engine(authenticated_url, pool_pre_ping=True)

with engine.begin() as connection:
    connection.execute(text(f"DROP TABLE IF EXISTS `{old_table_name}`"))

df.to_sql(
    table_name,
    con=engine,
    if_exists="replace",
    index=False,
)

with engine.connect() as connection:
    current_database = connection.execute(text("SELECT DATABASE()")).scalar_one()
    row_count = connection.execute(
        text(f"SELECT COUNT(*) FROM `{table_name}`")
    ).scalar_one()
    column_names = [
        row[0] for row in connection.execute(
            text(f"SHOW COLUMNS FROM `{table_name}`")
        ).fetchall()
    ]

print(f"Connected to MySQL database: {current_database}")
print(f"Table: {table_name}")
print(f"Uploaded {row_count} rows")
print(f"Columns: {column_names}")
```
